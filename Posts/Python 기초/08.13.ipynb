### 8장 1강 실습 문제 풀기 전 + 추가 내용

#### class = 설계도
#### class def = 매서드 설계도 안에 자동차를 어떻게 구성할 건지
#### 인스턴스 -> 설계도를 바탕으로 만들어진 자동차
#### self : 자동차 한 대(자기자신)
#### self.속성: 지금 이 자동차가 가져야 하는 정보(색, 4륜 ...)
#### init : 초기 설계안 표시
#### str : 객체를 print() 하면 어떤 것을 출력할지


```python
class Car:
    def __init__(self, color):
        self.color = color # 어떤 차가 올지 모르지만 오면 이 색깔
    def drive(self, speed):
        self.speed = speed
        print("이 자동차 빨라요")

car1=Car("red")
print(car1)

# 출력 내용 (잘 나온 것임)
#  <__main__.Car object at 0x0000024842951940> 
```

```python
class Car:
    def __init__(self, color):
        self.color = color # 어떤 차가 올지 모르지만 오면 이 색깔
    def drive(self, speed):
        self.speed = speed
        print("이 자동차 빨라요")
    def __str__(self):
        return f"{self.color} 자동차"

car1=Car("red")
print(car1)
```

```python
class Student:
    def __init__(self,name,score):
        self.name = name
        self.score = score
    def introduce(self):
        print(f"저는 {self.name}이고, 제 점수는 {self.score}입니다.")

student1 = Student("강민정",90)
student2 = Student("조형우", 20)
student1.introduce()
student2.introduce()

        
```
