### 5장 2강: 변수 스코프와 람다 함수

변수 스코프(Scope): 파이썬에서 변수를 선언할 때, 그 변수가 프로그램의 어느 장소까지 읽히고 수정될 수 있는지에 대한 '유효 범위'  

- **전역 변수(Global)** = 함수 바깥쪽에 선언되어 프로그램 파일 전체에서 언제든지 읽고 접근할 수 있는 변수  
ex) 모든 직원이 볼 수 있는 회사 공용 게시판  
- **지역 변수(Local)** = 특정 함수 내부 공간에 선언되어 오직 그 함수가 작동하는 동안에만 숨쉬는 변수  
ex) 특정 직원만 사용하는 개인 메모장  

```python
# 예시

# 전역 변수 선언
message = "저는 전역 변수입니다"

def test_scope():
    # 전역 변수와 이름이 똑같은 지역 변수 선언
    message = "저는 지역 변수입니다"
    print(f"[함수 내부 출력] {message}")  # 지역 변수가 우선 식별됨

test_scope()
print(f"[함수 외부 출력] {message}")  # 전역 변수 값은 오염되지 않고 유지됨
```

### 1.2 파이썬의 변수 검색 규칙: LEGB 스코프 개념

Local      → 현재 함수 안  
Enclosing  → 나를 감싸는 함수 안  
Global     → 파일 전체  
Built-in   → 파이썬 기본 공간  

```python
# 예시

name = "Global"

def outer():
    name = "Enclosing"

    def inner():
        name = "Local"
        print(name)

    inner()

outer()

# inner()에서 가장 가까운 Local 영역에 name이 있기 때문에 더 바깥쪽의 name은 찾지 않습니다.
```

**LEGB 네 가지 계층 구조**

---

변수를 찾아내는 우선순위는 다음의 4단계 범위를 따르며, 조건 만족 여부에 따라 첫 번째로 매핑되는 변수의 값을 도출합니다.  

1. **L (Local):** 현재 작동 중인 함수 내부 공간 (가장 높은 우선순위)  
2. **E (Enclosing):** 중첩 함수 구조에서 내부 함수를 감싸고 있는 외부 함수의 공간  
3. **G (Global):** 함수 바깥쪽이자 프로그램 소스코드 파일 전체의 전역 공간  
4. **B (Built-in):** 파이썬이 태생적으로 내장하고 있는 print(), len() 같은 기본 함수 공간 (가장 낮은 우선순위)  

```python
# 예시

# G (Global) 스코프 변수
value = "전역 데이터"

def outer_function():
    # E (Enclosing) 스코프 변수
    value = "외부 데이터"
    
    def inner_function():
        # L (Local) 스코프 변수
        value = "내부 데이터"
        print(value)  # 변수 식별 가동 -> Local에 있는 "내부 데이터"가 최종 도출됨
        
    inner_function()

outer_function()
```

```python
# 연습문제2.

#LEGB 규칙에 따라 변수를 찾고자 한다.
#다음 코드에서 `inner()` 함수가 `"외부 함수"`를 출력하도록 `outer()` 함수 내부에 `message` 변수를 선언하세요.

message = "전역"

def outer():
    message = "외부 함수"

    def inner():
        print(message)

    inner()

outer()
```

### 1.3 global 키워드의 작동 문법과 스코프 꼬임 현상 예방법  




```python
# 함수 안에서 단순히 전역 변수를 확인하는 것은 가능

visitor_count = 10

def show_count():
    print(visitor_count)
    
show_count()

# 하지만 함수 안에서 값을 대입하면 기본적으로 새로운 지역 변수를 다루게 됨.
# 전역 변수를 수정하겠다고 명시하려면 global을 사용.

visitor_count = 10

def add_visitor():
    global visitor_count
    visitor_count += 1

add_visitor()

print(visitor_count)

# global visitor_count : 지금부터 새 변수를 만드는 게 아니라 바깥의 visitor_count를 사용하겠습니다. 라고 파이썬에게 알려주는 것과 같음
```

**전역 변수 수정 통로 개방**

---

원래 파이썬 함수 내부에서는 전역 변수의 값을 읽을 수만 있고 수정하는 명령은 차단됩니다. 하지만 함수 내에서 **global 변수명**을 명시적으로 선언하면, 인터프리터에게 "지금부터 수정할 변수는 전역 공간에 있는 저 변수다"라고 식별 신호를 주어 값을 바꿀 수 있게 허용합니다.

**스코프 꼬임 현상 예방 가이드**

---

global 키워드를 무분별하게 남용하면 여러 함수가 하나의 전역 변수를 두고 서로 수치를 바꾸는 난잡한 구조가 됩니다. 이는 코드 실행 순서에 따라 결과가 계속 바뀌어 버그 진단이 불가능해지는 '**스코프 꼬임 현상**'을 유발하므로,  **함수의 인수를 통해 데이터를 입력받고 return으로 결과값을 돌려받는 독립적인 흐름 설계**를 취해야 합니다.

```python
counter = 0  # 전역 카운터 변수

def increase_counter():
    global counter  # 전역 변수 counter의 수정 통로를 개방
    counter += 1    # 전역 공간의 데이터가 차례대로 누적 증가함

increase_counter()
increase_counter()
print(f"최종 카운트 결과값: {counter}")  # 출력: 2
```
