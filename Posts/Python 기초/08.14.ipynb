```python
class PointWallet:
    def __init__(self, owner, points):
        self.owner = owner
        self.points = points

    def add_points(self, amount):
        self.points += amount
        print(f"{amount}포인트 적립")

    def use_points(self, amount):
        if self.points >= amount:
            self.points -= amount
            print(f"{amount}포인트 사용")
        else:
            print("포인트가 부족합니다.")

wallet = PointWallet("강민정", 5000)
# init을 실행 -> wallet 인스턴스 안에 owner, point 값이 각각 들어감


wallet.add_points(1000)
#인스턴스 안에 point 값에 10000 더해준 뒤 갱신 -> self.points이 6000이 됨
wallet.use_points(2000)
#인스턴스 안에 point 값에 20000 빼준 뒤 갱신 -> self.points이 4000이 됨

print(f"현재 포인트: {wallet.points}")
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner          # 예금주
        self.balance = balance      # 잔액

    # print()를 쓸 때 친숙하게 출력되도록 돕습니다.
    def __str__(self):
        return f"{self.owner}님의 계좌 (현재 잔액: {self.balance}원)"

    # 디버거 창이나 객체 단독 조회 시 명확하게 상태를 보여줍니다.
    def __repr__(self):
        return f"BankAccount(owner='{self.owner}', balance={self.balance})"

    def deposit(self, amount):
        self.balance += amount      # 입금 처리
        print(f"[{self.owner} 계좌] {amount}원 입금 / 현재 잔액: {self.balance}원")
        return self.balance

    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount  # 출금 처리
            print(f"[{self.owner} 계좌] {amount}원 출금 / 현재 잔액: {self.balance}원")
        else:
            print(f"[{self.owner} 계좌] 출금 실패: 잔액이 부족합니다. (잔액: {self.balance}원)")
        return self.balance
```

```python
class GameCharacter:
    def __init__(self, name, hp):
        self.name = name
        self.hp = hp

    def take_damage(self, damage):
        self.hp -= damage
        print(f"{self.name} 체력: {self.hp}")

warrior = GameCharacter("랜디", 100)
wizard = GameCharacter("배티", 80)

warrior.take_damage(30)

print(f"랜디 최종 체력: {warrior.hp}")
print(f"배티 최종 체력: {wizard.hp}")
```
