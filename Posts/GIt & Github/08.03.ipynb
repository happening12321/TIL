## 1장 1강: Git 동작 원리와 3영역 모델

workspace(작업 공간) → index(스테이징) → local repository(로컬 저장소) → remote repository(원격 저장소) 순서로 변경이 이동  

add: 작업 공간의 변경을 스테이징으로  
commit: 스테이징을 로컬 저장소로 올림  
push/pull: 로컬과 원격을 주고받음  

명령어	      이동 방향	                          하는 일
git status: (조회만)                             각 영역에 어떤 변경이 있는지 현재 상태를 보여줌
git add:    Working Directory → Staging Area	변경 파일을 다음 커밋 대상으로 올림
git commit: Staging Area → Repository	        스테이징된 변경을 스냅샷으로 영구 저장
